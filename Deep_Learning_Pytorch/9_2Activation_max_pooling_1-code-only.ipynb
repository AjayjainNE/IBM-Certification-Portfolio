```python
import torch 
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np
from scipy import ndimage, misc
```

```python
image=torch.zeros(1,1,5,5)
image[0,0,:,2]=1
image
```

```python
Z=conv(image)
Z
```
